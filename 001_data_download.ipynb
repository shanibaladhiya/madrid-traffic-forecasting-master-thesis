```python
import os
import re
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

BASE = "https://datos.madrid.es/dataset/300352-0-meteorologicos-horarios/downloads"
OUT_DIR = "madrid_weather_2023_csv"
os.makedirs(OUT_DIR, exist_ok=True)

html = requests.get(BASE, timeout=30).text
soup = BeautifulSoup(html, "html.parser")

links = []
for a in soup.find_all("a", href=True):
    text = " ".join(a.get_text(" ", strip=True).split())
    href = urljoin(BASE, a["href"])

    if (
        "2023" in text
        and "CSV Datos meteorológicos" in text
        and href.endswith("-csv")
    ):
        # Convert resource page URL to actual CSV download URL
        resource_id = href.rstrip("/").split("/")[-1]
        download_url = f"{href}/download/{resource_id}.csv"
        links.append((text, download_url))

print(f"Found {len(links)} CSV files")

for i, (title, url) in enumerate(links, start=1):
    filename = f"2023_{i:02d}.csv"
    path = os.path.join(OUT_DIR, filename)

    print("Downloading:", title)
    r = requests.get(url, timeout=60)
    r.raise_for_status()

    with open(path, "wb") as f:
        f.write(r.content)

print("Done.")
```

```python
import pandas as pd

weather = pd.read_csv(
    "madrid_weather_2023_2026_with_station_location.csv",
    low_memory=False
)

calendar = pd.read_csv(
    "CalendarData.csv",
    sep=";",
    low_memory=False
)

# Rename calendar columns to English
calendar.rename(
    columns={
        "Dia": "DATE",
        "Dia_semana": "DAY_OF_WEEK",
        "laborable / festivo / domingo festivo": "DAY_TYPE_RAW",
        "Tipo de Festivo": "HOLIDAY_TYPE",
        "Festividad": "HOLIDAY_NAME"
    },
    inplace=True
)

# Translate weekday names
weekday_map = {
    "lunes": "Monday",
    "martes": "Tuesday",
    "miercoles": "Wednesday",
    "miércoles": "Wednesday",
    "jueves": "Thursday",
    "viernes": "Friday",
    "sabado": "Saturday",
    "sábado": "Saturday",
    "domingo": "Sunday"
}

calendar["DAY_OF_WEEK"] = (
    calendar["DAY_OF_WEEK"]
    .str.lower()
    .map(weekday_map)
)

# Convert raw day type to clean English day type
daytype_map = {
    "laborable": "Weekday",
    "sabado": "Weekend",
    "sábado": "Weekend",
    "domingo": "Weekend",
    "festivo": "Holiday"
}

calendar["DAY_TYPE"] = (
    calendar["DAY_TYPE_RAW"]
    .str.lower()
    .map(daytype_map)
)

# Convert calendar date
calendar["DATE"] = pd.to_datetime(
    calendar["DATE"],
    format="%d/%m/%Y"
)

#only useful calendar columns
calendar = calendar[
    [
        "DATE",
        "DAY_OF_WEEK",
        "DAY_TYPE",
    ]
]

# Create DATE in weather file from YEAR, MONTH, DAY
weather["DATE"] = pd.to_datetime(
    dict(
        year=weather["YEAR"],
        month=weather["MONTH"],
        day=weather["DAY"]
    )
)

# Merge calendar info into weather file
weather = weather.merge(
    calendar,
    on="DATE",
    how="left"
)

# Move DATE and calendar columns near DAY
cols = weather.columns.tolist()

for col in ["DATE", "DAY_OF_WEEK", "DAY_TYPE"]:
    cols.remove(col)

day_index = cols.index("DAY")

cols[day_index + 1:day_index + 1] = [
    "DATE",
    "DAY_OF_WEEK",
    "DAY_TYPE"
]

weather = weather[cols]

# Save final file
weather.to_csv(
    "madrid_weather_2023_2026_with_stationLocation_calendar.csv",
    index=False
)

print("Saved successfully")
print(weather.shape)

print(
    weather.head(20)
)
```

```python
import os
import re
import time
import zipfile
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from pathlib import Path
from io import BytesIO

BASE = "https://datos.madrid.es/dataset/208627-0-transporte-ptomedida-historico/downloads"

START_YEAR = 2023
END_YEAR = 2026

month_map = {
    "Enero": "01",
    "Febrero": "02",
    "Marzo": "03",
    "Abril": "04",
    "Mayo": "05",
    "Junio": "06",
    "Julio": "07",
    "Agosto": "08",
    "Septiembre": "09",
    "Octubre": "10",
    "Noviembre": "11",
    "Diciembre": "12"
}

def get_with_retries(url, retries=5, sleep_seconds=10):
    for attempt in range(1, retries + 1):
        try:
            r = requests.get(url, timeout=180)
            r.raise_for_status()
            return r
        except requests.exceptions.RequestException as e:
            print(f"Attempt {attempt} failed:")
            print(e)

            if attempt == retries:
                raise

            print(f"Waiting {sleep_seconds} seconds before retry...")
            time.sleep(sleep_seconds)

html = get_with_retries(BASE).text
soup = BeautifulSoup(html, "html.parser")

resources = []

for a in soup.find_all("a", href=True):
    text = " ".join(a.get_text(" ", strip=True).split())
    href = urljoin(BASE, a["href"])

    match = re.search(
        r"ZIP Histórico de datos del tráfico\. (\w+) (\d{4})",
        text
    )

    if match and href.endswith("-zip"):
        month_name = match.group(1)
        year = int(match.group(2))

        if START_YEAR <= year <= END_YEAR:
            month_num = month_map.get(month_name)

            if month_num:
                resource_id = href.rstrip("/").split("/")[-1]
                download_url = f"{href}/download/{resource_id}"

                resources.append({
                    "year": year,
                    "month": month_num,
                    "title": text,
                    "url": download_url
                })

resources = sorted(resources, key=lambda x: (x["year"], x["month"]))

print(f"Found {len(resources)} ZIP files")

failed = []

for item in resources:
    year = item["year"]
    month = item["month"]

    out_dir = Path(f"madrid_traffic_{year}_csv")
    out_dir.mkdir(exist_ok=True)

    output_path = out_dir / f"{year}_{month}_traffic.csv"

    # Resume feature: skip if file already exists
    if output_path.exists() and output_path.stat().st_size > 0:
        print(f"Already exists, skipping: {output_path}")
        continue

    print(f"Downloading {year}-{month}: {item['title']}")

    try:
        response = get_with_retries(item["url"], retries=5, sleep_seconds=15)

        with zipfile.ZipFile(BytesIO(response.content)) as z:
            csv_files = [
                name for name in z.namelist()
                if name.lower().endswith(".csv")
            ]

            if not csv_files:
                print(f"No CSV found inside ZIP for {year}-{month}")
                failed.append(item)
                continue

            csv_name = csv_files[0]

            with z.open(csv_name) as source, open(output_path, "wb") as target:
                target.write(source.read())

            print(f"Saved: {output_path}")

        time.sleep(2)

    except Exception as e:
        print(f"Failed permanently for {year}-{month}")
        print(e)
        failed.append(item)

print("Done.")

if failed:
    print("\nFailed files:")
    for item in failed:
        print(item["year"], item["month"], item["url"])
else:
    print("All files downloaded successfully.")
```

```python
import time
import zipfile
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from pathlib import Path
from io import BytesIO

BASE = "https://datos.madrid.es/dataset/208627-0-transporte-ptomedida-historico/downloads"
YEAR = 2026

month_map = {
    "Enero": "01",
    "Febrero": "02",
    "Marzo": "03",
    "Abril": "04",
    "Mayo": "05",
    "Junio": "06",
    "Julio": "07",
    "Agosto": "08",
    "Septiembre": "09",
    "Octubre": "10",
    "Noviembre": "11",
    "Diciembre": "12"
}

def get_with_retries(url, retries=5):
    for attempt in range(1, retries + 1):
        try:
            r = requests.get(url, timeout=180)
            r.raise_for_status()
            return r
        except requests.exceptions.RequestException as e:
            print(f"Attempt {attempt} failed: {e}")
            time.sleep(10)
    raise Exception(f"Failed after {retries} retries: {url}")

html = get_with_retries(BASE).text
soup = BeautifulSoup(html, "html.parser")

resources = []

for a in soup.find_all("a", href=True):
    text = " ".join(a.get_text(" ", strip=True).split())
    href = urljoin(BASE, a["href"])

    if "Histórico de datos del tráfico" in text and str(YEAR) in text:
        for month_name, month_num in month_map.items():
            if month_name in text:
                resource_id = href.rstrip("/").split("/")[-1]
                download_url = f"{href}/download/{resource_id}.zip"

                resources.append({
                    "month": month_num,
                    "title": text,
                    "url": download_url
                })

resources = sorted(resources, key=lambda x: x["month"])

print(f"Found {len(resources)} files for {YEAR}")
for r in resources:
    print(r["month"], r["title"])
    print(r["url"])

out_dir = Path(f"madrid_traffic_{YEAR}_csv")
out_dir.mkdir(exist_ok=True)

for item in resources:
    output_path = out_dir / f"{YEAR}_{item['month']}_traffic.csv"

    if output_path.exists() and output_path.stat().st_size > 0:
        print(f"Already exists, skipping: {output_path}")
        continue

    print(f"Downloading {YEAR}-{item['month']}")

    response = get_with_retries(item["url"])

    with zipfile.ZipFile(BytesIO(response.content)) as z:
        csv_files = [
            name for name in z.namelist()
            if name.lower().endswith(".csv")
        ]

        if not csv_files:
            print("No CSV found:", item["title"])
            continue

        with z.open(csv_files[0]) as source, open(output_path, "wb") as target:
            target.write(source.read())

    print(f"Saved: {output_path}")

print("Done.")
```

```python
import pandas as pd
from pathlib import Path

output_path = Path.home() / "Downloads" / "weather_hourly.parquet"

hour_cols = [f"HOUR_{i:02d}" for i in range(1, 25)]
validity_cols = [f"VALIDITY_{i:02d}" for i in range(1, 25)]

id_cols = [
    "STATION_CODE",
    "STATION_NAME",
    "MAGNITUDE_NAME",
    "DATE",
    "LATITUDE",
    "LONGITUDE",
    "ALTITUDE"
]

value_long = weather.melt(
    id_vars=id_cols,
    value_vars=hour_cols,
    var_name="HOUR_COL",
    value_name="WEATHER_VALUE"
)

validity_long = weather.melt(
    id_vars=id_cols,
    value_vars=validity_cols,
    var_name="VALIDITY_COL",
    value_name="VALIDITY"
)

value_long["HOUR_NUM"] = value_long["HOUR_COL"].str.extract(r"(\d+)").astype(int)
validity_long["HOUR_NUM"] = validity_long["VALIDITY_COL"].str.extract(r"(\d+)").astype(int)

weather_long = value_long.merge(
    validity_long[id_cols + ["HOUR_NUM", "VALIDITY"]],
    on=id_cols + ["HOUR_NUM"],
    how="left"
)

weather_long["WEATHER_DATETIME"] = (
    pd.to_datetime(weather_long["DATE"]) +
    pd.to_timedelta(weather_long["HOUR_NUM"] - 1, unit="h")
)

weather_long = weather_long[weather_long["VALIDITY"] == "V"].copy()

weather_long["MAGNITUDE_NAME"] = (
    weather_long["MAGNITUDE_NAME"]
    .str.upper()
    .str.replace(" ", "_", regex=False)
)

weather_hourly = weather_long.pivot_table(
    index=[
        "STATION_CODE",
        "STATION_NAME",
        "WEATHER_DATETIME",
        "LATITUDE",
        "LONGITUDE",
        "ALTITUDE"
    ],
    columns="MAGNITUDE_NAME",
    values="WEATHER_VALUE",
    aggfunc="mean"
).reset_index()

weather_hourly = weather_hourly.rename(columns={
    "LATITUDE": "WEATHER_LATITUDE",
    "LONGITUDE": "WEATHER_LONGITUDE",
    "ALTITUDE": "WEATHER_ALTITUDE"
})

weather_hourly.to_parquet(output_path, index=False)

print("Saved:", output_path)
print(weather_hourly.shape)
print(weather_hourly.columns.tolist())
display(weather_hourly.head())
```
